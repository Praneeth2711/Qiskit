```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display
import matplotlib.pyplot as plt

simulator = AerSimulator()

print("Qiskit Challenge 2 initialized successfully!")
```

```python
import numpy as np

# Create a single-qubit circuit with one classical bit
qc_a = QuantumCircuit(1, 1)

# Apply an RY rotation of pi/3
qc_a.ry(np.pi / 3, 0)

# Measure the qubit
qc_a.measure(0, 0)

# Display the circuit
print("Circuit A: Coin biased toward 0")
display(qc_a.draw("mpl"))

# Execute for 100 shots
result_a = simulator.run(qc_a, shots=100).result()

# Get measurement counts
counts_a = result_a.get_counts()

print("Measurement results:", counts_a)
```

```python
import matplotlib.pyplot as plt
from qiskit.visualization import plot_histogram
```

```python
# Circuit A: Histogram

print("Circuit A measurement results:", counts_a)

plot_histogram(
    counts_a,
    title="Circuit A - Quantum Coin Biased Toward 0",
    figsize=(8, 5)
)

plt.show()
```

```python
import matplotlib.pyplot as plt

zero_count = counts_a.get('0', 0)
one_count = counts_a.get('1', 0)

plt.figure(figsize=(8, 5))
plt.bar(['0', '1'], [zero_count, one_count])

plt.title("Circuit A - Quantum Coin Biased Toward 0")
plt.xlabel("Measurement Outcome")
plt.ylabel("Number of Shots")
plt.ylim(0, 100)

plt.show()
```

```python
import numpy as np
from qiskit import QuantumCircuit
from IPython.display import display

# Create a quantum circuit with 1 qubit and 1 classical bit
qc_b = QuantumCircuit(1, 1)

# Rotate the qubit to favor outcome 1
qc_b.ry(2 * np.pi / 3, 0)

# Measure the qubit
qc_b.measure(0, 0)

# Display the circuit
display(qc_b.draw("mpl"))

# Run the circuit 100 times
result_b = simulator.run(qc_b, shots=100).result()

# Get measurement counts
counts_b = result_b.get_counts()

print("Circuit B measurement results:", counts_b)
```

```python
import matplotlib.pyplot as plt

zero_count_b = counts_b.get('0', 0)
one_count_b = counts_b.get('1', 0)

plt.figure(figsize=(8, 5))
plt.bar(['0', '1'], [zero_count_b, one_count_b])

plt.title("Circuit B - Quantum Coin Biased Toward 1")
plt.xlabel("Measurement Outcome")
plt.ylabel("Number of Shots")
plt.ylim(0, 100)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

outcomes = ['0', '1']

# Get counts for Circuit A
counts_a_values = [
    counts_a.get('0', 0),
    counts_a.get('1', 0)
]

# Get counts for Circuit B
counts_b_values = [
    counts_b.get('0', 0),
    counts_b.get('1', 0)
]

x = np.arange(len(outcomes))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(x - width/2, counts_a_values, width, label='Circuit A (favors 0)')
plt.bar(x + width/2, counts_b_values, width, label='Circuit B (favors 1)')

plt.xticks(x, outcomes)
plt.xlabel("Measurement Outcome")
plt.ylabel("Number of Shots")
plt.title("Comparison of Biased Quantum Coins")
plt.ylim(0, 100)
plt.legend()
plt.show()
```
