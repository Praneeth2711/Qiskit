```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from IPython.display import display
import matplotlib.pyplot as plt
import pandas as pd

print("Libraries imported successfully!")
```

```python
# Circuit A: One X gate

qc_a = QuantumCircuit(1, 1)

qc_a.x(0)

qc_a.measure(0, 0)

display(qc_a.draw("mpl"))
```

```python
simulator = AerSimulator()

result_a = simulator.run(qc_a, shots=100).result()

counts_a = result_a.get_counts()

print("Circuit A results:", counts_a)
```

```python
# Circuit B: Two X gates

qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)

qc_b.measure(0, 0)

display(qc_b.draw("mpl"))
```

```python
result_b = simulator.run(qc_b, shots=100).result()

counts_b = result_b.get_counts()

print("Circuit B results:", counts_b)
```

```python
# Circuit C: One Hadamard gate

qc_c = QuantumCircuit(1, 1)

qc_c.h(0)

qc_c.measure(0, 0)

display(qc_c.draw("mpl"))
```

```python
result_c = simulator.run(qc_c, shots=100).result()

counts_c = result_c.get_counts()

print("Circuit C results:", counts_c)
```

```python
print("Circuit A:", counts_a)
print("Circuit B:", counts_b)
print("Circuit C:", counts_c)
```

```python
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

print("Histogram libraries imported successfully!")
```

```python
import matplotlib.pyplot as plt

results = [counts_a, counts_b, counts_c]

titles = [
    "Circuit A: One X Gate",
    "Circuit B: Two X Gates",
    "Circuit C: One H Gate"
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, counts, title in zip(axes, results, titles):
    states = ["0", "1"]
    values = [counts.get("0", 0), counts.get("1", 0)]

    ax.bar(states, values)
    ax.set_title(title)
    ax.set_xlabel("Measurement Outcome")
    ax.set_ylabel("Number of Shots")
    ax.set_ylim(0, 110)
    ax.set_xticks([0, 1], ["0", "1"])

plt.tight_layout()
plt.show()
```
